```
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

* We have imported every required libraries.

* We will read csv file with the help of pandas.

```
df = pd.read_csv('Diwali Sales Data.csv', encoding='unicode_escape')
#to avoid encoding error, use 'unicode_escape'
```

```
df.shape
```

```
df.head(10)
```

```
df.info()
```

* We can see that status and unnamed has no non null value so we dont need them ...


* so we will drop them by using drop function.....


* Ammount also have 12 null values...

```
df.drop(['Status','unnamed1'], axis=1, inplace=True)
```

```
df.head(10)
```

```
df.info()
```

* Now we will check for total null values in the table

```
pd.isnull(df).sum()
```

* Now we have 12 null values in ammounts so we will delete that 12 rows from the data

```
df.dropna(inplace=True)
```

```
df.shape
```

```
#change data type;
df['Amount'] = df['Amount'].astype('int')
```

```
df.info()
```

```
df.columns
```

```
df.rename(columns={'Marital_Status': 'Shadi'})
```

```
df.describe()
```

```
df[['Age','Orders','Amount']].describe()
```

Exploratory Data Analysis

Gender

```
df.columns
```

```
ax = sns.countplot(x='Gender', data=df)

#for marking value at the bars
for bars in ax.containers:
    ax.bar_label(bars)
```

```
sales = df.groupby(['Gender'],as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)
```

```
ax = sns.barplot(x = 'Gender', y = 'Amount', data = sales)

#for marking value at the bars
for container in ax.containers:
    ax.bar_label(container)
```

* From the above graph we can see that most the buyers are fenmales and even the purchasing power of females are greater than males.

Age

```
df.columns
```

```
ax = sns.countplot(data = df, x = 'Age Group', hue = 'Gender')

for bars in ax.containers:
    ax.bar_label(bars)
```

```
age = df.groupby(['Age Group'], as_index = False)['Amount'].sum().sort_values(by='Amount', ascending=False)

sns.barplot(x = 'Age Group', y = 'Amount', data = age)
```

* From above graph we can see that most of the buyers are of age group between 26-35 yrs female

State

```
df.columns
```

```
state = df.groupby(['State'], as_index= False)['Orders'].sum().sort_values(by='Orders', ascending=False).head(10)

sns.set(rc={'figure.figsize':(20,5)})
sns.barplot(data=state, x='State', y='Orders')
```

```
state = df.groupby(['State'], as_index= False)['Amount'].sum().sort_values(by='Amount', ascending=False).head(10)

sns.set(rc={'figure.figsize':(20,5)})
sns.barplot(data=state, x='State', y='Amount')
```

* We can see that most of the orders are ordered from UP Maharashtra and karnatka and also most amount is spend by these 3.

Marital Status

```
df.columns
```

```
ax = sns.countplot(x = 'Marital_Status', data = df)

sns.set(rc={'figure.figsize':(5,5)})
for bars in ax.containers:
    ax.bar_label(bars)
```

```
Shadi = df.groupby(['Marital_Status','Gender'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)
sns.barplot(x = 'Marital_Status', y = 'Amount', data = Shadi,  hue = 'Gender')
```

* From the above graph we can see that most of the buyers are married women and they also have higher purchasing power.

Occupation

```
df.columns
```

```
sns.set(rc={'figure.figsize':(20,5)})
ax = sns.countplot(x = 'Occupation', data = df)
for bars in ax.containers:
    ax.bar_label(bars)
```

```
occupation = df.groupby(['Occupation'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)

sns.barplot(x = 'Occupation', y = 'Amount', data = occupation)
```

* From above graph we can see that most of the buyers are working in IT, Aviation and Healthcare sector

Product Category


```
df.columns
```

```
df.head(10)
```

```
sns.set(rc={'figure.figsize':(30,5)})
ax = sns.countplot(x = 'Product_Category', data = df)

for bars in ax.containers:
    ax.bar_label(bars)
```

```
product = df.groupby(['Product_Category'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False).head(10)
sns.set(rc={'figure.figsize':(20,5)})
sns.barplot(x = 'Product_Category', y = 'Amount', data = product)
```

* From the above graph we can see that most of the sold products are from food, Clothing & Apparel and electronic category

```
id = df.groupby(['Product_ID'], as_index=False)['Orders'].sum().sort_values(by='Orders', ascending=False).head(10)
sns.barplot(x = 'Product_ID', y = 'Orders', data = id)
```

Conclusion:

* Married women age group 26-35 yrs from UP, Maharashttra and Karnataka working in IT, Healthcare and Aviation are more likely to buy products like Food Clothing and
